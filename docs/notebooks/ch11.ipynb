## 11.1 Derived classes -- using inherited members

Once a derived class is built, an instance can use the base class's attributes and methods as if they were its own.

```python
class Account:
    def __init__(self):
        self.owner = ''
        self.balance = 0

    def display(self):
        print(self.owner, self.balance)

class SavingsAccount(Account):
    def __init__(self):
        Account.__init__(self)
        self.rate = 0.0

# TODO: create a SavingsAccount instance, set its owner and balance to
#       values of your choice, then call the inherited display() method
```

## 11.2 Accessing base class attributes

Python searches the instance, then its class, then each base class up the inheritance tree, and stops at the first match.

```python
class Employee:
    def __init__(self, name):
        self.name = name

    def info(self):
        print(f'{self.name} works here')

class Manager(Employee):
    def __init__(self, name, team_size):
        Employee.__init__(self, name)
        self.team_size = team_size

    def assign(self):
        print(f'{self.name} is assigning work to {self.team_size} people')

# TODO: define a class named Director derived from Manager whose __init__
#       takes name and team_size and passes both up to its base class --
#       then create a Director and call both assign() and info() on it
```

## 11.3 Overriding class methods -- replacing a method

An override is a method in a derived class with the same name as one in the base class. Because the search starts at the instance's class, the override wins.

```python
class Invoice:
    def __init__(self, amount):
        self.amount = amount

    def display(self):
        print(f'Invoice: {self.amount}')

# TODO: define a class named RushInvoice derived from Invoice that overrides
#       display() to print a line beginning with 'RUSH Invoice:' followed by
#       the amount -- then create one of each class and call display() on both
```

## 11.3 Overriding class methods -- polymorphism

When each class has its own `display()`, one loop can handle a mix of them and each object runs its own version.

```python
class Invoice:
    def __init__(self, amount):
        self.amount = amount

    def display(self):
        print(f'Invoice: {self.amount}')

class RushInvoice(Invoice):
    def display(self):
        print(f'RUSH Invoice: {self.amount * 1.25}')

# TODO: build a list containing both Invoice and RushInvoice instances, then
#       loop over it calling display() on each item
```

## 11.5 Mixin classes and multiple inheritance -- building a mixin

A mixin is a small class that adds a few methods to whatever uses it. It relies on attributes the class it's mixed into provides.

```python
class Report:
    def __init__(self, title, rows):
        self.title = title
        self.rows = rows

class EmailMixin:
    def send(self):
        print(f'Emailing {self.title}')

# TODO: define a mixin class named ExportMixin with a to_csv method that
#       prints the title and rows separated by a comma
```

## 11.5 Mixin classes and multiple inheritance -- combining base classes

Listing several base classes in the parentheses is multiple inheritance. Python searches them left to right.

```python
class Report:
    def __init__(self, title, rows):
        self.title = title
        self.rows = rows

class ExportMixin:
    def to_csv(self):
        print(f'{self.title},{self.rows}')

class EmailMixin:
    def send(self):
        print(f'Emailing {self.title}')

# TODO: define a class named SalesReport that inherits from Report and both
#       mixins -- then create a SalesReport and call both to_csv() and send()
```
