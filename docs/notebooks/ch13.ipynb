## 13.5 Reloading modules

A module only runs once per import, so edits to the file are not noticed. The cell below creates `rate_card.py` and imports it.

```python
from importlib import reload

with open('rate_card.py', 'w') as f:
    f.write('RATE = 0.2\n')

import rate_card

print(rate_card.RATE)

# TODO: rewrite rate_card.py so RATE = 0.25, then make the already imported
#       module pick up the change without restarting, and print rate_card.RATE.
```
