## 10.7 Class interfaces

The class interface is the set of methods a caller is meant to use. A leading underscore on a method name (like `_diff_time`) signals internal plumbing that isn't part of that interface -- Python doesn't enforce it, but it tells a reader not to depend on it directly.

```python
class RaceTime:
    def __init__(self, start_hrs, start_mins, end_hrs, end_mins, dist):
        self.start_hrs = start_hrs
        self.start_mins = start_mins
        self.end_hrs = end_hrs
        self.end_mins = end_mins
        self.distance = dist

    def print_pace(self):
        total_time = self._diff_time()
        total_minutes = total_time[0] * 60 + total_time[1]
        print(f'Avg pace (mins/mile): {total_minutes / self.distance:.2f}')

    # TODO: define an internal method named _diff_time that returns a
    #       2-tuple of (hours, minutes) representing end minus start times
    #       -- lead its name with an underscore to signal it's internal
    #       plumbing, not part of the class's interface -- then create a
    #       RaceTime instance of your choice and call print_pace() on it
```

## 10.9 More operator overloading: Classes as numeric types -- arithmetic operators

Arithmetic operators can be overloaded too. `time1 - time2` calls `time1.__sub__(time2)`, and the method can hand back a brand-new instance holding the result.

```python
class Time24:
    def __init__(self, hours, minutes):
        self.hours = hours
        self.minutes = minutes

    def __str__(self):
        return f'{self.hours:02d}:{self.minutes:02d}'

    def __gt__(self, other):
        if self.hours > other.hours:
            return True
        return self.hours == other.hours and self.minutes > other.minutes

    # TODO: define a __sub__ method that returns a new Time24 holding the
    #       absolute time difference between self and other (whichever is
    #       later, minus whichever is earlier) -- then create two Time24
    #       instances and print their difference
```
